```python
import tkinter as tk
from tkinter import messagebox

def place_bet():
    amount = amount_var.get()

    try:
        amount = float(amount)
        if amount <= 0:
            raise ValueError

        messagebox.showinfo(
            "Bet placed",
            f"Bet placed successfully for {amount:.2f} INR"
        )
    except ValueError:
        messagebox.showerror(
            "Invalid amount",
            "Please enter a valid positive amount."
        )

root = tk.Tk()
root.title("Betting Interface")
root.geometry("320x420")
root.configure(bg="#17191c")

tk.Label(
    root,
    text="UFC Official Partners",
    fg="#ff1744",
    bg="#17191c",
    font=("Arial", 16, "bold")
).pack(pady=25)

amount_var = tk.StringVar(value="10.00")

tk.Label(
    root,
    text="Bet amount (INR)",
    fg="white",
    bg="#17191c"
).pack()

tk.Entry(
    root,
    textvariable=amount_var,
    justify="center",
    font=("Arial", 14)
).pack(pady=10)

tk.Button(
    root,
    text="Bet10.00 INR",
    command=place_bet,
    bg="#16b500",
    fg="white",
    font=("Arial", 14),
    width=15,
    height=2
).pack(pady=20)

tk.Label(
    root,
    text="Demo interface only — no real money is involved.",
    fg="#aaaaaa",
    bg="#17191c",
    wraplength=260
).pack(pady=20)

root.mainloop()
```

```python
import random
import time

def aviator_round(balance, bet_amount):
    """
    Simulates one round of an Aviator-style game.
    - balance: current player balance
    - bet_amount: how much the player bets
    """
    if bet_amount > balance:
        print("Not enough balance!")
        return balance

    print(f"\nStarting round... Bet placed: {bet_amount}")
    balance -= bet_amount

    # Randomly generate a crash multiplier
    crash_point = round(random.uniform(2.0, 15.0), 2)

    multiplier = 2.0
    while multiplier < crash_point:
        print(f"Multiplier: {multiplier:.2f}x")
        time.sleep(0.3)  # simulate time passing
        multiplier += 0.25

        # Example auto‑cashout at 2x
        if multiplier >= 2.0:
            print(f"Auto cashout at {multiplier:.2f}x ✅")
            winnings = bet_amount * multiplier
            balance += winnings
            print(f"You won {winnings:.2f}! New balance: {balance:.2f}")
            return balance

    print(f"💥 Crash at {crash_point:.2f}x! You lost your bet.")
    print(f"Balance left: {balance:.2f}")
    return balance


# --- Run the simulation ---
balance = 1000.0
for i in range(3):
    balance = aviator_round(balance, 100.0)
```
