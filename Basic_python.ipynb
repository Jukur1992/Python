```python
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()  # move to next line
```

```python
from turtle import *
bgcolor("black")
setposition(-4, -2)
color('aqua')
tracer(3,0)
hideturtle()
for i in range(500):
    rt(i)
    circle(150,i)
    fd(70)
    right(210)
    fd(i)
    lt(1)

done()
```

```python
n=5
for i in range(n):
    upper=True
    for j in range(i+n):
        if upper:
            print(chr(65+j),end=' ')
        else:
            print(chr(97+j),end=' ') 
        upper=not upper
    print()    
```
