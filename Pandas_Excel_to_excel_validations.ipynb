'''Data  comparison, count check, null check, duplicate check. '''

```python
import pandas as pd

# Load both Excel files
df1 = pd.read_excel(r"D:\PYTHON-SQL\Python\Employee_source.xlsx")
df2 = pd.read_excel(r"D:\PYTHON-SQL\Python\Employee_target.xlsx")

# ✅ Step 1: Compare row counts
if len(df1) == len(df2):
    print("Row counts match")
else:
    print(f"Row count mismatch: File1={len(df1)}, File2={len(df2)}")

# ✅ Step 2: Compare column names
if list(df1.columns) == list(df2.columns):
    print("Column names match")
else:
    print("Column mismatch")
    print("File1 columns:", df1.columns)
    print("File2 columns:", df2.columns)

# ✅ Step 3: Find mismatched rows
diff = df1.compare(df2)
if diff.empty:
    print("No differences found ✅")
else:
    print("Differences found:")
    print(diff)

# ✅ Step 4: Optional – check duplicates
duplicates = df1[df1.duplicated()]
if not duplicates.empty:
    print("Duplicates in File1:")
    print(duplicates)
```
